```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from sklearn.preprocessing import MinMaxScaler
```

```python
# Define the path to the dataset
dataset_path = '/kaggle/input/spacenet-an-optimally-distributed-astronomy-data/'

# List all image files in the dataset
image_files = [f for f in os.listdir(dataset_path) if f.endswith('.png')]

# Select a random image
image_path = os.path.join(dataset_path, np.random.choice(image_files))

# Open the image
image = Image.open(image_path)

# Convert to grayscale
image_gray = image.convert('L')

# Normalize the image
scaler = MinMaxScaler()
image_array = np.array(image_gray)
image_normalized = scaler.fit_transform(image_array.reshape(-1, 1)).reshape(image_array.shape)

# Display the image
plt.imshow(image_normalized, cmap='gray')
plt.axis('off')
plt.show()
```
